```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
data = {
    "Student_ID": range(1, 21),

    "Study_Hours": [2, 5, 3, 7, 4, 6, 1, 8, 5, 3,
                    7, 2, 6, 4, 9, 3, 5, 8, 2, 6],

    "Attendance": [65, 85, 72, 95, 80, 88, 60, 97, 82, 70,
                   92, 68, 90, 78, 98, 75, 84, 94, 63, 87],

    "Assignment_Score": [55, 82, 68, 91, 76, 85, 48, 94, 80, 65,
                         89, 58, 86, 73, 96, 70, 81, 92, 52, 84],

    "Previous_Marks": [50, 78, 65, 88, 72, 82, 45, 91, 76, 60,
                       85, 54, 80, 70, 94, 66, 79, 89, 48, 81],

    "Final_Marks": [52, 80, 67, 92, 75, 86, 47, 95, 81, 64,
                    90, 57, 84, 74, 97, 69, 82, 93, 50, 85]
}

df = pd.DataFrame(data)

print("Dataset created successfully!")
```

## 1. Understanding the Dataset

The dataset contains information about 20 students and their academic performance.

It includes the following information:

- **Student ID:** Unique identification number for each student.
- **Study Hours:** Number of hours the student studies.
- **Attendance:** Student attendance percentage.
- **Assignment Score:** Score obtained in assignments.
- **Previous Marks:** Marks obtained in previous examinations.
- **Final Marks:** Marks obtained in the final examination.

This information will be analyzed to understand the factors related to student performance.

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
average_marks = df["Final_Marks"].mean()

print("Average Final Marks:", average_marks)
```

```python
top_student = df.loc[df["Final_Marks"].idxmax()]

print("Student with the highest final marks:")
print(top_student)
```

```python
low_student = df.loc[df["Final_Marks"].idxmin()]

print("Student with the lowest final marks:")
print(low_student)
```

```python
average_study_hours = df["Study_Hours"].mean()

print("Average Study Hours:", average_study_hours)
```

```python
average_attendance = df["Attendance"].mean()

print("Average Attendance:", average_attendance, "%")
```

```python
average_assignment_score = df["Assignment_Score"].mean()

print("Average Assignment Score:", average_assignment_score)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Study_Hours"], df["Final_Marks"])

plt.xlabel("Study Hours")
plt.ylabel("Final Marks")
plt.title("Study Hours vs Final Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Attendance"], df["Final_Marks"])

plt.xlabel("Attendance (%)")
plt.ylabel("Final Marks")
plt.title("Attendance vs Final Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Assignment_Score"], df["Final_Marks"])

plt.xlabel("Assignment Score")
plt.ylabel("Final Marks")
plt.title("Assignment Score vs Final Marks")

plt.show()
```

## 2. Analysis Findings

The visualizations help us understand the relationship between different
student factors and final academic performance.

The graphs indicate that study hours, attendance, and assignment performance
show an increasing pattern with final marks in this dataset.

```python
correlations = df[[
    "Study_Hours",
    "Attendance",
    "Assignment_Score",
    "Final_Marks"
]].corr()

print(correlations["Final_Marks"].sort_values(ascending=False))
```

```python
plt.figure(figsize=(8, 5))

plt.imshow(correlations, cmap="coolwarm")
plt.colorbar()

plt.xticks(range(len(correlations.columns)), correlations.columns, rotation=45)
plt.yticks(range(len(correlations.columns)), correlations.columns)

plt.title("Correlation Heatmap")
plt.tight_layout()

plt.show()
```

```python
average_values = [
    df["Study_Hours"].mean(),
    df["Attendance"].mean(),
    df["Assignment_Score"].mean(),
    df["Final_Marks"].mean()
]

labels = [
    "Study Hours",
    "Attendance",
    "Assignment Score",
    "Final Marks"
]

plt.figure(figsize=(9, 5))

plt.bar(labels, average_values)

plt.xlabel("Performance Factors")
plt.ylabel("Average Value")
plt.title("Average Student Performance Factors")

plt.xticks(rotation=20)
plt.show()
```

## 3. Key Findings

Based on the analysis of the student dataset, the following observations were made:

- The average final mark of the students is **76**.
- The average study time is **4.8 hours per day**.
- The average attendance is **81.5%**.
- The average assignment score is **76.25**.
- Students with higher study hours generally achieved higher final marks.
- Higher attendance generally showed a positive relationship with final marks.
- Assignment performance also showed a strong relationship with final marks.
- Student 15 achieved the highest final mark of **97**, while Student 7 achieved the lowest mark of **47**.

These findings help demonstrate how Python can be used to analyze student performance and identify patterns in academic data.

```python
def performance_category(marks):
    if marks >= 85:
        return "Excellent"
    elif marks >= 70:
        return "Good"
    else:
        return "Needs Improvement"


df["Performance_Category"] = df["Final_Marks"].apply(performance_category)

df[["Student_ID", "Final_Marks", "Performance_Category"]]
```

```python
category_counts = df["Performance_Category"].value_counts()

print(category_counts)
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    category_counts,
    labels=category_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Performance Categories")
plt.show()
```

```python
df.to_csv("student_performance_analysis.csv", index=False)

print("Final dataset saved successfully!")
```
